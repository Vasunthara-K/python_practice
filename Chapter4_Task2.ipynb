# Task 2. Merge Function 1 Example

```python
import pandas as pd

customer_details = pd.DataFrame({
    "customer_id": [101, 102, 103, 104],
    "name": ["Ravi", "Priya", "Akil", "Anu"]})
customer_details
```

```python
order_details = pd.DataFrame({
    "customer_id": [101, 102, 101, 105],
    "product": ["Laptop", "Phone", "Mouse", "Keyboard"],
    "amount": [50000, 20000, 1000, 3000]})
order_details
```

## Inner Merge

```python
inner_merge = pd.merge(customer_details,order_details,on="customer_id",how="inner")
inner_merge
```

```python
pd.merge(customer_details,order_details,on="customer_id")
```

## Left Merge

```python
Left_merge = pd.merge(customer_details,order_details,on="customer_id",how="left")
Left_merge
```

## Right Merge

```python
Right_merge = pd.merge(customer_details,order_details,on="customer_id",how="right")
Right_merge
```

## Outer Merge

```python
outer_merge = pd.merge(customer_details,order_details,on="customer_id",how="outer")
outer_merge
```

## Cross Merge - Every possible combination

```python
cross_merge = pd.merge(customer_details,order_details,how="cross")
cross_merge
```

# 1. Create a new list containing each number multiplied by 5

```python
numbers = [2, 4, 6, 8, 10]
new_numbers =[i * 5 for i in numbers]
new_numbers
```

# 2. Create a new list containing only the numbers greater than 10

```python
numbers = [2, 5, 8, 11, 14, 17, 20]
new_list = [i for i in numbers if i > 10]
new_list
```

```python
numbers = [2, 5, 8, 11, 14, 17, 20]
new_list = [ i * i for i in numbers if i > 10]
new_list
```

# Task 4: Lambda with Map,Filter,Reduce

```python
numbers = [1, 2, 3, 4, 5]
square = list(map(lambda x: x * x, numbers))
square
```

## 1. create a new list where every number is multiplied by 3.

```python
numbers = [2, 4, 6, 8, 10]
new_list = list(map(lambda x: x * 3 , numbers))
new_list
```

## 2. Convert all names to uppercase.

```python
names = ["ravi", "priya", "akil", "anu"]
upper_case_names = list(map(lambda x: x.upper(),names))
upper_case_names
```

## 3. keep only numbers greater than 15.

```python
numbers = [5, 10, 15, 20, 25, 30]
new_numbers = list(filter(lambda x: x > 15,numbers))
new_numbers
```

## 4. keep only adults (age 18 and above)

```python
ages = [12, 18, 25, 16, 30, 15, 21]
new_list = list(filter(lambda x: x > 18,ages))
new_list
```

## 5. calculate the sum of all numbers.

```python
from functools import reduce
numbers = [2, 4, 6, 8]
total = reduce(lambda x,y: x + y,numbers)
total
```

## 6. calculate the multiplication of all numbers.

```python
numbers = [2, 3, 4, 5]
product = reduce(lambda x,y: x*y,numbers)
product
```
