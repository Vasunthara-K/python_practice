## step 1: Handle Missing values(NaN)

```python
amazon_sales_viz_data["discounted_price"] = amazon_sales_viz_data["discounted_price"].str.replace("₹","")
```

```python
amazon_sales_viz_data["actual_price"] = amazon_sales_viz_data["actual_price"].str.replace("₹","")
```

```python
amazon_sales_viz_data["discounted_price"] = amazon_sales_viz_data["discounted_price"].str.replace(",","")
```

```python
amazon_sales_viz_data["actual_price"] = amazon_sales_viz_data["actual_price"].str.replace(",","")
```

```python
amazon_sales_viz_data["discount_percentage"] = amazon_sales_viz_data["discount_percentage"].str.replace("%","")
```

```python
amazon_sales_viz_data["rating_count"] = amazon_sales_viz_data["rating_count"].str.replace(",","")
```

```python
amazon_sales_viz_data["discounted_price"] = pd.to_numeric(amazon_sales_viz_data["discounted_price"])
```

```python
amazon_sales_viz_data["actual_price"] = pd.to_numeric(amazon_sales_viz_data["actual_price"])
```

```python
amazon_sales_viz_data["discount_percentage"] = pd.to_numeric(amazon_sales_viz_data["discount_percentage"])
```

```python
amazon_sales_viz_data["rating_count"] = pd.to_numeric(amazon_sales_viz_data["rating_count"])
```

```python
amazon_sales_viz_data["rating"] = pd.to_numeric(amazon_sales_viz_data["rating"])
```

# Data Visualization

```python
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
amazon_sales_viz_data["category_broad"].value_counts()
```

```python
sns.countplot(data=amazon_sales_viz_data["category_broad"])
plt.title("No.of product in each category")
plt.xlabel("product_count")
plt.ylabel("product_category")
plt.show()
```

```python
amazon_sales_viz_data["rating"].describe()
```

```python
plt.hist(amazon_sales_viz_data["rating"])
plt.title("Distribution of Product Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Products")
plt.show()
```

```python
sns.barplot(data=amazon_sales_viz_data,x="discount_percentage",y="category_broad")
plt.title("Discount Percentage by Product Category")
plt.xlabel("Discount Percentage")
plt.ylabel("Product Category")
plt.show()
```

## How are actual price and discounted price related?
This is a relationship between two numerical columns, so we'll use a scatter plot.

```python
sns.scatterplot(x=amazon_sales_viz_data["actual_price"],y=amazon_sales_viz_data["discounted_price"])
plt.title("Actual Price vs Discounted Price")
plt.xlabel("Actual Price")
plt.ylabel("Discounted Price")
plt.show()
```

# Discount Percentage vs Product Rating
### Does discount percentage have a relationship with product rating

```python
sns.scatterplot(data=amazon_sales_viz_data,x="rating",y="discount_percentage")
plt.title("Discount Percentage vs Product Rating")
plt.xlabel("Product Rating")
plt.ylabel("Discount Percentage (%)")
plt.show()
```

# Distribution of Discounted Prices

## What relationships exist between product pricing, discounts, ratings, and customer engagement?

```python
correlation = amazon_sales_viz_data [["actual_price",
     "discounted_price",
     "discount_percentage",
     "rating",
     "rating_count"]].corr().round(2)

correlation
```
