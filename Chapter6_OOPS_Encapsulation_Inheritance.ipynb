# 1. Encapsulation

```python
class bankaccount:
    def __init__(self,__balance):
        self.__balance = __balance
        print(self.__balance)

account = bankaccount(10000)
account.__balance= -20000
print(account.__balance)

#account.balance = -20000
```

# 2. Inheritance

```python
class bankaccount:
    
    def __init__(self,name,balance):
        self.name = name
        self.balance = balance
    
    def view_balance(self):
        print(self.name)
        print("current balance is:",self.balance)
    
    def deposit(self,amount):
        self.balance= amount + self.balance
        print(self.name)
        print("after deposit balance is:",self.balance)

class savingsaccount(bankaccount):
    pass

account1 = bankaccount("account1",1000)
account1.view_balance()

account1.deposit(200)
#account1.view_balance()

account2 = savingsaccount("account2",200)
account2.deposit(400)
#account2.view_balance()

        
```

## step1: lets add method in child class
* Child class can use the parent's __init__

```python
class BankAccount:
    def __init__(self,balance):
        self.balance = balance
    
    def view_balance(self):
        print("Showing balance",self.balance)

class SavingsAccount(BankAccount):
    def add_interest(self):
        print("interest added")

account = SavingsAccount(100)
account.view_balance()    # from parent class

account3 = SavingsAccount(200)
account3.add_interest()  # from child class
```

# step 2 - Method Override

* Method overriding = child class replaces the parent's method with its own method.

```python
class bankaccount:
    def __init__(self,balance):
        self.balance = balance

    def view_balance(self):
        print("displaying parent method balance:")

class savingsaccount(bankaccount):
    
    def view_balance(self):
        print("displaying child method balance")

account = savingsaccount(100)  #child view balance, uses init from parent so required 100
account.view_balance() # even though parent has view balance child override parant method

account1 = bankaccount(100)
account1.view_balance()
```

## step 3 - child class can also have init

```python
class bankaccount:
    def __init__(self,balance):
        self.balance = balance

    def view_balance(self):
        print("displaying parent method balance:")

class savingsaccount(bankaccount): # child class can also have init. if child has init it wont inherit parent's __init__
    def __init__(self,balance,interest):
        self.balance = balance
        self.interest = interest  # self.balance = balance will be inherited from parent class
    
    def view_balance(self):
        print("displaying child method balance")

    def add_interest(self):
        self.balance = self.balance + self.interest
        print("interest is:",self.balance)

account = savingsaccount(100,50)
account.add_interest()
```

## step 4 - super()

* super().method_name() → call the parent's version of that method, then continue with the child method.
* Run the parent's __init__() also, Go to the parent's __init__()
* Execute the parent's code
* Come back to the child's __init__()
* Continue with the next child statement
* super().__init__(), super().view_balance()

```python
class bankaccount:
    def __init__(self,balance):
        self.balance = balance
        print("parent init:",self.balance)

class savingsaccount(bankaccount):
    def __init__(self,balance):
        super().__init__(balance) # calling parents constructor from child
        print("child init :",self.balance)
        #self.interest = interest

    def add_interest(self,interest):
        self.balance = self.balance + interest
        print("balance is:",self.balance)

account = savingsaccount(100)
account.add_interest(50)
    
```

```python
class bankaccount:
    def __init__(self,balance):
        self.balance = balance

    def view_balance(self):
        print("parent balance:",self.balance)

class savingsaccount(bankaccount):
    def __init__(self,total):
        self.total = total
        super().__init__(total) # calling parents constructor

    def view_balance(self):
        super().view_balance()
        print("child total :",self.total)
        #self.interest = interest

    def add_interest(self,interest):
        self.balance = self.balance + interest
        print("balance is:",self.balance)

account = savingsaccount(100)
account.view_balance()
account.add_interest(50)
```
