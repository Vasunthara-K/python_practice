## Step 2: Polymorphism with inheritance

* A child class can inherit from a parent.
* A child can override a parent's method.
* super() can call the parent's version.

```python
class bankaccount:
    def __init__(self,balance):
        self.balance = balance

    def view_balance(self):
        print("balance from parent method:",self.balance)

class quarter_interest(bankaccount):

    def add_interest(self,interest):
        print("quarter interest is:",self.balance/4)

    def view_balance(self):
        super().view_balance()
        print("balance from child method:",self.balance)

#account1 = bankaccount(1000)
#account1.view_balance()


account2 = quarter_interest(200)
account2.view_balance()
account2.add_interest(100)
    
    
```

## polymorphism with for loop

```python
class bankaccount:
    def __init__(self,balance):
        self.balance = balance

    def view_balance(self):
        print("balance from parent method:",self.balance)

class quarter_interest:

    def add_interest(self,interest):
        self.interest = interest
        print("quarter interest is:",self.interest/4)

class annual_interest:
    def add_interest(self,interest):
        self.interest = interest
        print("Annual interest is:",self.interest)

account1 = quarter_interest() # explicitly define class for the object
account2 = annual_interest() # explicitly define class for the object

objects = [account1,account2] #store objects in the list

for i in objects:
    i.add_interest(100)
```

# Abstraction

* Abstraction means hiding unnecessary implementation details and exposing only what the user needs to use.

## step1 Abstract class

```python
from abc import ABC, abstractmethod # From Python's abc module, bring me ABC and abstractmethod so I can create an abstract class

class Model(ABC): # Here Model becomes an abstract base class.but can't create an object of an abstract class

    @abstractmethod # This method(predict) must be implemented by the child class but can have its own functionality
    def predict(self):
        pass

class LinearModel(Model):
    def predict(self):
        print("Linear prediction")

class RegressionModel(Model):
    def predict(self):
        print("RegressionModel prediction")

model1 = LinearModel()
model1.predict()

model2 = RegressionModel()
model2.predict()
```
