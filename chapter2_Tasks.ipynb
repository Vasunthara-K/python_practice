# 1. List[]

```python
#1.Count how many customers are 31 years old.
customer_ages = [25, 31, 22, 31, 40,28, 22, 35, 31, 27]
customer_ages.count(31)
```

```python
#2. Add a new customer aged 29
customer_ages = [25, 31, 22, 31, 40,28, 22, 35, 31, 27]
customer_ages.append(29)
customer_ages
```

```python
#Add 2 new customers aged 35
customer_ages = [25, 31, 22, 31, 40,28, 22, 35, 31, 27]
customer_ages.insert(1,35)
customer_ages
```

```python
#3. Add 2 new customers aged 21,23 from set ages
customer_ages = [25, 31, 22, 31, 40,28, 22, 35, 31, 27]
ages ={21,23}
customer_ages.extend(ages)
customer_ages
```

```python
#4. Remove one customer aged 22.
customer_ages = [25, 31, 22, 31, 40,28, 22, 35, 31, 27]
customer_ages.remove(22)
customer_ages
```

```python
#5. Sort the ages
customer_ages = [25, 31, 22, 31, 40,28, 22, 35, 31, 27]
customer_ages.sort()
customer_ages
```

```python
#6. Reverse the sorted list.
customer_ages = [25, 31, 22, 31, 40,28, 22, 35, 31, 27]
customer_ages.sort(reverse=True)
customer_ages
```

```python
#7. Find the index of age 40
customer_ages = [25, 31, 22, 31, 40,28, 22, 35, 31, 27]
customer_ages.index(40)
```

```python
#8. Copy the list
customer_ages = [25, 31, 22, 31, 40,28, 22, 35, 31, 27]
copied_customer_ages = customer_ages.copy()
copied_customer_ages
```

```python
#9. Clear the copied list.
customer_ages = [25, 31, 22, 31, 40,28, 22, 35, 31, 27]
copied_customer_ages = [25, 31, 22, 31, 40,28, 22, 35, 31, 27]
copied_customer_ages.clear()
copied_customer_ages
```

```python
#10 Print the maximum, minimum, sum, and average age
customer_ages = [25, 31, 22, 31, 40,28, 22, 35, 31, 27]
print(min(customer_ages))
print(max(customer_ages))
print(sum(customer_ages))
average_age = sum(customer_ages)/len(customer_ages)
print((average_age))
```

```python
# all unique employee IDs
team_A = {"emp_101", "emp_102", "emp_103", "emp_104", "emp_105"}
team_B = {"emp_104", "emp_105", "emp_106", "emp_107", "emp_108"}
team_C = team_A.symmetric_difference(team_B)
team_C
```

```python
# employees working only in team A
team_A = {"emp_101", "emp_102", "emp_103", "emp_104", "emp_105"}
team_B = {"emp_104", "emp_105", "emp_106", "emp_107", "emp_108"}
team_C = team_A.difference(team_B)
team_C
```

```python
#employees working in both teams
team_A = {"emp_101", "emp_102", "emp_103", "emp_104", "emp_105"}
team_B = {"emp_104", "emp_105", "emp_106", "emp_107", "emp_108"}
team_C = team_A.intersection(team_B)
team_C
```

```python
# employees only in Team A
team_A = {"emp_101", "emp_102", "emp_103", "emp_104", "emp_105"}
team_B = {"emp_104", "emp_105", "emp_106", "emp_107", "emp_108"}
team_A
```

```python
# all employess in both the team.
team_A = {"emp_101", "emp_102", "emp_103", "emp_104", "emp_105"}
team_B = {"emp_104", "emp_105", "emp_106", "emp_107", "emp_108"}
team_C = team_A.union(team_B)
team_C
```

```python
team_A = {"emp_101", "emp_102", "emp_103", "emp_104", "emp_105"}
team_B = {"emp_104", "emp_105", "emp_106", "emp_107", "emp_108"}
team_A.issubset(team_B)
```

```python
team_A = {"emp_104", "emp_105"}
team_B = {"emp_104", "emp_105", "emp_106", "emp_107", "emp_108"}
team_A.issubset(team_B)
```

```python
team_A = {"emp_104", "emp_105"}
team_B = {"emp_104", "emp_105", "emp_106", "emp_107", "emp_108"}
team_A.issuperset(team_B)
```

```python
team_A = {"emp_104", "emp_105"}
team_B = {"emp_104", "emp_105", "emp_106", "emp_107", "emp_108"}
team_A.isdisjoint(team_B)
```
