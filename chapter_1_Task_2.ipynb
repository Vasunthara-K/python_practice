# Example 1 : multiplication table

```python
user_input = int(input("Enter the number:"))
start_table = 1
while start_table <= 5:
    multiplication = user_input * start_table
    print("{} * {} = {}".format(start_table,user_input,multiplication))
    start_table = start_table + 1
```

# Example 1 : Find vowels present in given word

```python
user_input = input("Enter the word to find vowels present:")
vowels = "a", "e", "i", "o", "u"
count = 0
for i in user_input:
    if i.lower() in vowels :
        count += 1
        print(i)
print("count of vowels in given word is:",count)
```

# Example 1 : student grade system

```python
try:
    total_mark = int(input("Please enter total mark of the student:"))
    
    if total_mark > 100 or total_mark < 1:
        print("Mark should be from 1 to 100")
    elif total_mark >= 90 and total_mark <= 100:
        print("Grade 'A'")
    elif total_mark >= 70 and total_mark < 90:
        print("Grade 'B'")
    elif total_mark >= 50 and total_mark < 70:
        print("Grade 'C'")
    else:
        print("Need to improve")
except:
    print("only numbers from 1 to 100 allowed")
```

# Example 2: Find square root every number until user give number

```python
def square_root(number):
    for i in range(1,number+1):
        square = i*i
        print(f"square root of {i} is:",square) 
square_root(3)
```

# Example 1: find even numbers until user given number

```python
def even_number():
    user_number = int(input("Enter the number:"))
    for i in range(1,user_number+1):
        if i % 2 == 0:
            print(i) 
       
even_number()
```

# Example 2: counts the number of spaces in string

```python
def count_space(string):
    count = 0
    for i in string:
        if i == " ":
            count += 1
    print(count)
        
count_space("AI Learning with handson") 
```

# Example 1 : enable the option to enter the password 3 times

```python
root = "admin"
attempt=1
while True:
    user_password = input("please enter your password:")
    
    if user_password == root:
        print("**Login Successful**")
        break
    elif attempt <=2 :
        attempt +=1
        print("Password didnt match")
    else:
        print("Trial ends")
        break
        
```

# Example 2 : Print numbers from 1 to 30, skipping all multiples of 5

```python
for i in range(1,21):
    if i % 5 == 0:
        continue
    elif i % 10 == 0:
        break
    else:
        print(i)
        
```
