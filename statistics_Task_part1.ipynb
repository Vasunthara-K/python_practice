# Question 4
## Historically, 40% of applicants pass an entrance test. After a new preparation program, 90 out of 180 applicants pass. At the 5% level of significance, test whether the pass rate has improved.

## Test : one sample proportion test(right tailed)
## H0(Null Hypothesis) p = 0.40 : 40% of applicants pass an entrance test
## H1(Alternate Hypothesis) p > 0.40 : the pass rate has improved

```python
40/100
```

```python
90/180
```

```python
population_proportion = 0.40
sample_size = 180
passed_applicants = 90
alpha = 0.05
```

```python
sample_proportion = passed_applicants / sample_size

print(sample_proportion)
```

```python
import math

# SE tells us about the expected variation of sample proportions if we repeatedly took samples of 180 applicants.
standard_error = math.sqrt(
    population_proportion * (1 - population_proportion) / sample_size
)

z_statistic = (sample_proportion - population_proportion) / standard_error

print("Standard Error:", standard_error)
print("z-statistic:", z_statistic)
```

```python
from scipy.stats import norm

p_value = norm.sf(z_statistic)

print("p-value:", p_value)
```

```python
if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

## At the 5% significance level, there is sufficient statistical evidence to conclude that the pass rate under the new program is increased

# Question 5

## Two teaching methods are used to train employees. A sample of 30 employees trained using Method A has a mean score of 75 with a standard deviation of 12. A sample of 28 employees trained using Method B has a mean score of 70 with a standard deviation of 10. At the 5% level of significance, test whether there is a difference in average scores between the two methods.

## Test : Two sample T-test (two tailed)
## H0(Null Hypothesis) : μ1 = μ2:  there is no significant difference in average scores between the two methods.
## H1(Alternate Hypothesis) : μ1 != μ2 there is a significant difference between Two teaching methodss.

```python
# Method A
n1 = 30
mean1 = 75
std1 = 12

# Method B
n2 = 28
mean2 = 70
std2 = 10

# Significance level
alpha = 0.05
```

```python
import math

standard_error = math.sqrt(
    (std1**2 / n1) + (std2**2 / n2)
)

print("Standard Error:", standard_error)
```

```python
t_statistic = (mean1 - mean2) / standard_error

print("t-statistic:", t_statistic)
```

```python
df = (
    (std1**2 / n1 + std2**2 / n2)**2
    /
    (
        ((std1**2 / n1)**2 / (n1 - 1))
        +
        ((std2**2 / n2)**2 / (n2 - 1))
    )
)

print("Degrees of freedom:", df)
```

```python
from scipy.stats import t

p_value = 2 * t.sf(abs(t_statistic), df)

print("p-value:", p_value)
```

```python
if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

## At the 5% significance level, there is insufficient statistical evidence to conclude that the average scores differ between Method A and Method B.

# Question 6

## The average delivery time from two warehouses is compared using random samples. Warehouse A has a sample of 35 deliveries with a mean time of 48 minutes and a standard deviation of 6 minutes. Warehouse B has a sample of 40 deliveries with a mean time of 52 minutes and a standard deviation of 7 minutes. At the 5% level of significance, test whether the mean delivery times differ.

## Test : Two sample t-test(Two-tailed)
## H0(Null Hypothesis) : μ1 = μ2 : No significant difference in mean delivery time between the two warehouses
## H1(Alternate Hypothesis) : μ1 != μ2 : There is significant difference in delivery time between Warehouse A and Warehouse B

```python
# Warehouse A
n1 = 35          # Number of deliveries in Warehouse A
mean1 = 48       # Average delivery time for Warehouse A
std1 = 6         # Standard deviation of delivery times in Warehouse A

# Warehouse B
n2 = 40          # Number of deliveries in Warehouse B
mean2 = 52       # Average delivery time for Warehouse B
std2 = 7         # Standard deviation of delivery times in Warehouse B

alpha = 0.05     # Significance level = 5%
```

```python
import math

standard_error = math.sqrt((std1**2 / n1) + (std2**2 / n2))  # SE of the difference between the two sample means

print("Standard Error:", standard_error)  # Shows how much the difference between sample means can naturally vary
```

```python
t_statistic = (mean1 - mean2) / standard_error  # Difference between the two sample means divided by its SE

print("t-statistic:", t_statistic)  # Shows how many standard errors the observed mean difference is from 0
```

```python
df = ((std1**2/n1 + std2**2/n2)**2) / (((std1**2/n1)**2/(n1-1)) + ((std2**2/n2)**2/(n2-1)))  # Welch's degrees of freedom

print("Degrees of freedom:", df)  # Used to determine the t-distribution for the test
```

```python
from scipy import stats

p_value = 2 * stats.t.sf(abs(t_statistic), df)  # Two-tailed p-value: considers both directions of difference

print("p-value:", p_value)  # Probability of observing a difference this large if H0 is true
```

```python
if p_value < alpha:  # If p-value is less than 5% significance level
    print("Reject H0")
else:
    print("Fail to reject H0")
```

## At the 5% level of significance, we can conclude that there is a statistical evidence of difference in mean delivery time between warhouse A and warehouse B
